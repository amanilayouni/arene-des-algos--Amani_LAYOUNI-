```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target)

print(X.shape, y.shape)
print("Classes :", data.target_names)
```

```python
# phase 1
from sklearn.model_selection import train_test_split

def split_train_val_test(X, y, test_size=0.2, val_size=0.2, random_state=42):



    if val_size <= 0:
        raise ValueError("val_size doit être strictement positif (sinon pas de validation set)")

    # 1. SPLIT TEST
    X_temp, X_test, y_temp, y_test = train_test_split(
        X,
        y,
        test_size=test_size,
        stratify=y,
        random_state=random_state
    )

    # 2. SPLIT VALIDATION SUR LE RESTE

    val_ratio = val_size / (1 - test_size)

    X_train, X_val, y_train, y_val = train_test_split(
        X_temp,
        y_temp,
        test_size=val_ratio,
        stratify=y_temp,
        random_state=random_state
    )

    # 3. AFFICHAGE CONTRÔLE

    print("Train :", len(X_train),
          "| Validation :", len(X_val),
          "| Test :", len(X_test))

    # Vérification des classes
    print("\nRépartition des classes :")
    print("Train :", y_train.value_counts(normalize=True).values)
    print("Val   :", y_val.value_counts(normalize=True).values)
    print("Test  :", y_test.value_counts(normalize=True).values)

    return X_train, X_val, X_test, y_train, y_val, y_test
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = split_train_val_test(X, y)
```

```python
try:
    split_train_val_test(X, y, val_size=0)
except Exception as e:
    print("Erreur contrôlée :", e)
```

```python
import numpy as np

y_imbalanced = y.copy()
y_imbalanced[:int(len(y)*0.95)] = 1
y_imbalanced[int(len(y)*0.95):] = 0

split_train_val_test(X, y_imbalanced)
```
